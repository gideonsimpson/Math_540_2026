# Convergence Testing without an Exact Solution
Test problem is
$$
u' = u^2 - t, \quad 0<t\leq 1,\quad u(0) = 1
$$
Reference solution is with $h=10^{-6}$

```julia
N_vals = [10,100, 1000, 10000, 100000, 1000000];
Nref = 10^8;
h_vals = 1 ./ N_vals;

# set common parameters
f= (t,u) -> u^2 - t;
u0 = 1.0;
t0 = 0.0;
tmax = 1.0;

t_refs, u_refs = explicit_euler(f, u0, t0, tmax, Nref);
u_ref = u_refs[end]; # store reference solution at tmax

errors = zeros(length(N_vals));
for (i, N) in enumerate(N_vals)
    t_vals, u_vals = explicit_euler(f, u0, t0, tmax, N);
    # compute error at tmax against the reference solution
    errors[i] = abs(u_vals[end] - u_ref);
end

errors
```

Errror is diminishing

```julia
scatter(h_vals, errors, xscale=:log10, yscale=:log10, marker=:o, label="Error at tmax"
     , xlabel="h", ylabel="Error", title="Explicit Euler Error vs Step Size", legend=:topleft)
plot!(h_vals, 100*h_vals, label="O(h)", linestyle=:dash)
xticks!(10.0.^(-6:1:0))
yticks!(10.0.^(-4:1:2))
```

This empirical evidence that there is $\mathrm{O}(h)$ convergence.

# Vector Valued Problems
This solves the Lotka-Volterra predator prey system:
\begin{align}
u_1' & =u_1 -.1 u_1 u_2\\
u_2' & = -1.5 u_2 + 0.075 u_1 u_2
\end{align}

```julia
#common parameters and choices for Lokta-Volterra system
function f_lk(t, u)
    rhs = zeros(2);
    rhs[1] = u[1] - .1 * u[1] * u[2];
    rhs[2] = -1.5 * u[2] + .075 * u[1]*u[2];
    return rhs;
end

u0 = [2.0, 1.0];
t0 = 0.0
tmax = 100.0;
N = 100000;

# solve
t_vals, u_vals = explicit_euler(f_lk, u0, t0, tmax, N);
```

Returns an array of arrays/vector of vectors

```julia
u_vals
```

```julia
# plot first component (prey population) over time from vector of vectors
plot(t_vals, [u_[1] for u_ in u_vals], label="Prey", xlabel="Time", ylabel="Population", title="Lotka-Volterra System")
plot!(t_vals, [u_[2] for u_ in u_vals], label="Predator")
```

Look at ``phase'' portrait:

```julia
plot([u_[1] for u_ in u_vals], [u_[2] for u_ in u_vals], label="Phase Plane", xlabel="Prey Population", ylabel="Predator Population", title="Lotka-Volterra Phase Plane")
```

We are supposed to see closed loops; the error of the numerical method introduces a ``wobble''
