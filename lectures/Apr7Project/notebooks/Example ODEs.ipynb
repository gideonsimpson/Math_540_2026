# Example 1
Solve
$$
u' = 0, \quad u(0)= 1
$$

```julia
f(t,u) = 0.0
u0 = 1.0
t0 = 0.0;
tmax = 10.0;
N = 100;
t_vals, u_vals = explicit_euler(f, u0, t0, tmax, N);
```

# Example 2
$$
u' = -u, \quad u(0) = 1.0
$$
Recall that the answer is $u(t) = e^{-t}$.

```julia
f(t,u) = -u;
u0 = 1.0
t0 = 0.0;
tmax = 10.0;
N = 100;
t_vals, u_vals = explicit_euler(f, u0, t0, tmax, N);
```

```julia
plot(t_vals, u_vals, label="Explicit Euler", xlabel="t", ylabel="u(t)", title="ODE Solution")
plot!(t_vals, exp.(-t_vals), label="Exact Solution", linestyle=:dash)
```

# Example 3
$$
u' = -10 \cdot u, \quad u(0) = 1.0
$$
Recall that the answer is $u(t) = e^{-10 t}$.


```julia
f(t,u) = -10 * u;
u0 = 1.0
t0 = 0.0;
tmax = 10.0;
N = 10;
t_vals, u_vals = explicit_euler(f, u0, t0, tmax, N);

plot(t_vals, u_vals, label="Explicit Euler", xlabel="t", ylabel="u(t)", title="ODE Solution")
plot!(t_vals, exp.(-10 * t_vals), label="Exact Solution", linestyle=:dash)
```

For this problem, if the step size is too large, we get a numerical instability.  If you go back and use a larger `N` (smaller `h`), this will correct itself.

# Detour: Exploring undef

```julia
n = 100
x = Vector{Float64}(undef, n); # undef instructs the computer to just allocate memory without initializing it
```

```julia
y = zeros(n) # zeros initializes all entries to 0.0
```

Why bother?

```julia
using BenchmarkTools
```

Time the computational cost of allocating arrays in these two ways:

```julia
n = 10000;

@btime x = Vector{Float64}(undef, n);
@btime y = zeros(n);
@btime z = ones(n);
@btime w = rand(n);
```

For fastest code, try to allocate yur memory up front.

# Implicit Euler Example

```julia
f(t,u) = -10 * u;
u0 = 1.0
t0 = 0.0;
tmax = 10.0;
N = 10;
t_vals, u_vals = implicit_euler(f, u0, t0, tmax, N);

plot(t_vals, u_vals, label="Explicit Euler", xlabel="t", ylabel="u(t)", title="ODE Solution")
plot!(t_vals, exp.(-10 * t_vals), label="Exact Solution", linestyle=:dash)
```

Implicit Euler has no issues with large step sizes, at least for this problem.
