# Boundary Value Problem Solver

## Example 1
Solve
$$
-u''=2, \quad u(0) = u(1) = 0
$$

```julia
function rhs!(du, u, p, x)
    du[1] = u[2];
    du[2] = -2;
    du
end

function bc!(res, u, p, x)
   ga, gb = p;

    res[1] = u(x[1])[1] - ga;
    res[2] = u(x[end])[1] - gb;
    res
end
```

```julia
xspan = (0, 1.0)
p = (0.0, 0.0)
```

Starting guess for $\mathbf{u} = (u_1, u_2)^T$; try a constant function

```julia
uguess = [0.0, 0.0];
```

```julia
bvp = BVProblem(rhs!, bc!, uguess, xspan, p)
```

```julia
soln = solve(bvp)
```

```julia
xx = LinRange(xspan[1], xspan[2], 100)
uex = -xx.^2 + xx;

plot(soln, vars=(1), label="Numerical Solution", title="BVP Solution", xlabel="x", ylabel="u(x)")
plot!(xx, uex, label="Exact Solution")
```

## Neumann Condition
Solve
$$
-u''=2, \quad u(0) =0, \quad u'(1) = 1
$$

```julia
function rhs!(du, u, p, x)
    du[1] = u[2];
    du[2] = -2;
    du
end

function bc!(res, u, p, x)
   ga, hb = p;

    res[1] = u(x[1])[1] - ga;
    res[2] = u(x[end])[2] - hb;
    res
end
```

```julia
xspan = (0, 1.0)
p = (0.0, 1.0)
uguess = [0.0, 0.0];
bvp = BVProblem(rhs!, bc!, uguess, xspan, p)
soln = solve(bvp)
```

```julia
plot(soln, vars = (1), label="Numerical Solution", title="BVP Solution with Mixed BCs", xlabel="x", ylabel="u(x)")
```

## Nonlinear

```julia
function rhs!(du, u, p, x)
    du[1] = u[2];
    du[2] = -u[1]^2;
    du
end

function bc!(res, u, p, x)
   ga, gb = p;

    res[1] = u(x[1])[1] - ga;
    res[2] = u(x[end])[1] - gb;
    res
end

xspan = (0, 1.0)
p = (0.0, 0.0)
uguess = [0.0, 0.0];
bvp = BVProblem(rhs!, bc!, uguess, xspan, p)
soln = solve(bvp)

```

```julia
plot(soln, vars=(0, 1)  )
```

To get a nontrivial solution in the nonlienar setting, you might need to finesse your initial guess

```julia
using Plots
```

```julia
# ------------------------------------------------------------
# Fourier-Lagrange basis function ℓ_j(x)
# N must be odd: N = 2n+1
# ------------------------------------------------------------
function fourier_lagrange_basis(j, x, N)

    xj = 2π * j / N
    θ = x - xj
    # abvoid small divsors
    if abs(sin(θ/2)) < 1e-14
        return 1.0
    else
        return sinc(N * θ/(2*π))/sinc(θ/(2*π))
    end
end
```

```julia
n = 8
N = 2 * n + 1
x = LinRange(0, 2π, 500)
x_nodes = 2π * (0:N-1) / N


plot(x, fourier_lagrange_basis.(0, x, N), label="ℓ₀")
plot!(x, fourier_lagrange_basis.(1, x, N), label="ℓ₁")
plot!(x, fourier_lagrange_basis.(2, x, N), label="ℓ₂")
plot!(x_nodes, ones(N), ls=:dash, color = :black, label="")
scatter!(x_nodes, zeros(N), label="")
xlabel!("x")
```
