```julia
using LinearAlgebra
using Plots
using BoundaryValueDiffEq
```

Solve $- u'' = 2$ with $u(0) = 0$ and $u(1) = 0$.

```julia
function f!(du, u, p, t)
    du[1] = u[2]
    du[2] =  -2.0;
    du
end

function bc!(res, u, p, t)
    a,b = p
    # left endpoint
    res[1] = u(a)[1] - 0.0
    # right endpoint
    # res[2] = u(end)[1] - 0.0
    res[2] = u(b)[1] - 0.0

    res
end
```

```julia
uguess = [0.0, 0.0]
a = 0.
b = 1.

p = (a, b)
tspan = (a, b)
bvp = BVProblem(f!, bc!, [0.0, 0.0], tspan, p)
```

```julia
soln = solve(bvp);
plot(soln, vars=(1), label="BVP Solution", xlabel="t", ylabel="u(t)")
xx = range(a, b, length=100)
plot!(xx, -xx.^2 .+ xx, label="Exact Solution", linestyle=:dash)
```
