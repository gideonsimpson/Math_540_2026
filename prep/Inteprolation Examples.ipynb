```julia
using Plots
using Interpolations
```

```julia
x = LinRange(0, 2*π, 10)
y = @. sin(x);

itp_linear = linear_interpolation(x, y)
itp_cubic = cubic_spline_interpolation(x, y)

x_dense = LinRange(0, 2*π, 100)
plot(x_dense, itp_linear(x_dense), label="Linear Interpolant")
plot!(x_dense, itp_cubic(x_dense), label="Cubic Interpolant")
plot!(x_dense, sin.(x_dense), label="sin(x)", ls=:dash)
scatter!(x, y, label="Data")
xlabel!("x")
savefig("interpolation_plot.pdf")
```

```julia
# number of sample points
N = 32

# periodic grid: include repeated point at 2π
x = LinRange(0, 2π, N+1);
y = @. sin(x) + 0.5 * cos(5*x);

# set up the periodic interpolant on our mesh
periodic_interpolant_ = interpolate(y, BSpline(Cubic(Periodic(OnGrid()))))
# relate this to the grid and enable periodic extrapolation outside [minimum(x), maximum(x)]
periodic_interpolant = extrapolate(scale(periodic_interpolant_, x), Periodic());

# evaluate the interpolant on a dense grid beyond where we have data
x_dense = LinRange(-2π, 4π, 500)
plot(x_dense, periodic_interpolant.(x_dense), label="Periodic Cubic Interpolant")
plot!(x_dense, sin.(x_dense) + 0.5*cos.(5*x_dense), label="sin(x) + 0.5*cos(5x)", ls=:dash)
scatter!(x, y, label="Data")
xlabel!("x")
savefig("periodic_interpolation_plot.pdf")
```

```julia
# number of sample points
N = 16

# periodic grid:
# IMPORTANT: do not include both 0 and 2π
x = range(0, 2π, length=N+1)[1:end-1]

# sample the periodic function
f(x) = sin(x) + 0.3*cos(3x)

y = f.(x)

# construct periodic cubic spline interpolant
itp = interpolate(
    y,
    BSpline(Cubic(Periodic(OnGrid())))
)

# attach physical coordinates
sitp = scale(itp, x)

# fine grid for visualization
xfine = range(0, 2π, length=1000)

# exact and interpolated values
yexact = f.(xfine)
yinterp = sitp.(xfine)
```
