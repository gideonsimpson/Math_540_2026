```julia
using Plots
using Random
using LinearAlgebra
using Statistics
```

```julia
function harmonic_mc(g, x0;
                     Nsamples = 10_000,
                     Δt = 1e-4)

    vals = zeros(Nsamples)

    d = length(x0);

    for k in 1:Nsamples

        x = copy(x0)

        while norm(x) < 1.0
            x .+= sqrt(Δt) * randn(d)
        end

        vals[k] = g(x)

    end

    return mean(vals), std(vals) / sqrt(Nsamples)
end
```

```julia
function g1(x)
    θ = atan(x[2], x[1])   # θ such that tan(θ) = x[2]/x[1], with correct quadrant
    return cos(2θ)
end
```

```julia
harmonic_mc(g1, [0.0, 0.0])
```

```julia
x_ = LinRange(-1.0, 1.0, 25)
y_ = LinRange(-1.0, 1.0, 25)

Ugrid = fill(NaN, length(x_), length(y_))
for (i, x) in enumerate(x_), (j, y) in enumerate(y_)
    if x^2 + y^2 < 1.0
        Ugrid[i, j] = harmonic_mc(g1, [x, y], Nsamples=1000)[1]
    end
end 

surface(
    x_, y_, Ugrid',
    xlabel = "x", ylabel = "y", zlabel = "u",
    title = "Monte Carlo estimate on unit disk (surface)",
    c = :viridis
    )
```

```julia
# True solution: u(x, y) = x^2 - y^2 (shown on the unit disk)
Utrue = fill(NaN, length(x_), length(y_))
for (i, x) in enumerate(x_), (j, y) in enumerate(y_)
    if x^2 + y^2 < 1.0
        Utrue[i, j] = x^2 - y^2
    end
end

surface(
    x_, y_, Utrue',
    xlabel = "x", ylabel = "y", zlabel = "u",
    title = "True solution on unit disk: u = x^2 - y^2",
    c = :viridis
)
```
